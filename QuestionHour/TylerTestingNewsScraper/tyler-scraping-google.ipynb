```python
import requests
import re

url = "https://news.google.com/home?hl=en-US&gl=US&ceid=US:en"

headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/137.0.0.0 Safari/537.36"
    )
}

html = requests.get(url, headers=headers).text

print(html[:1000])
```

```python
matches = re.findall(
    r'aria-label="([^"]+)".*?</a>.*?role="presentation"',
    html,
    re.DOTALL
)

print(matches[:10])
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
import re
import time

options = Options()

driver = webdriver.Chrome(options=options)

driver.get(
    "https://news.google.com/home?hl=en-US&gl=US&ceid=US:en"
)

time.sleep(5)

html = driver.page_source

driver.quit()
```

```python
matches = re.findall(
    r'aria-label="([^"]+)".*?</a>.*?role="presentation"',
    html,
    re.DOTALL
)

for m in matches:
    print(m)
```

```python
import re

nav_items = re.findall(
    r'track:click">\s*([A-Za-z][A-Za-z\s&]+?)\s*<svg',
    html
)

print(nav_items)
```

```python
headers = re.findall(
    r'<h2[^>]*class="GTpP4b"[^>]*>(.*?)</h2>',
    html
)

print(headers)
```

```python
sections = {
    "nav": re.findall(r'track:click">\s*([A-Za-z][A-Za-z\s&]+?)\s*<svg', html),
    "headers": re.findall(r'<h2[^>]*class="GTpP4b"[^>]*>(.*?)</h2>', html),
    "subtext": re.findall(r'<p>(Recommended[^<]+)</p>', html)
}

print(sections)
```
